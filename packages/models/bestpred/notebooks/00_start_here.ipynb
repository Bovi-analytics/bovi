# 00 - Start Here: BESTPRED in Bovi

This is the entry point for a practical review of `bestpred-py`, the pure-Python
port of USDA BESTPRED. Work through the numbered notebooks in order on a first
visit. They separate the recommended application API from compatibility formats,
optional Farm Data Definitions (FDD), Bovi's current milk-only implementation,
and the external Fortran oracle.

The notebooks describe the current state; they do **not** imply that Bovi's
existing `lactationcurve` production paths have already been replaced.

## Environment

From the Bovi repository root, run:

```bash
just sync
uv run jupyter notebook packages/models/bestpred/notebooks
```

Python 3.12 and the Bovi uv workspace are the supported environment. The wheel
contains the runtime package, but this repository-oriented notebook series is not
included in the wheel.

````python
from pathlib import Path


def find_repo_root() -> Path:
    for candidate in (Path.cwd(), *Path.cwd().parents):
        if (candidate / "packages/models/bestpred").is_dir():
            return candidate
    raise RuntimeError("Run this notebook from a Bovi repository checkout")


ROOT = find_repo_root()
PACKAGE_ROOT = ROOT / "packages/models/bestpred"
FIXTURES = PACKAGE_ROOT / "tests/fixtures"
PARAMETERS = FIXTURES / "source11_current/bestpred.par"
````
